```python
import pandas as pd

datos = {
    "Nombre": ["Abril", "Alejandra", "Adan", "Nicole", "Karla"],
    "Calificacion": [9, 8, 10, 7, 9]
}

# DataFrame tabla o Marco de datos
df = pd.DataFrame(datos)

print("=== CALIFICACIONES ===")
# Imprimir informacion de datos
print(df)

print("\n Promedio y maximo ")
print("Promedio:", df["Calificacion"].mean())
print("Calificacion mayor:", df["Calificacion"].max())
print("-+-+-+-+-+-+-+-+-Jupyter Alejandra-+-+-+-+-+-+-+-")
print("-+-+-+-+-+-+-+-+-Alejandra Rivera Valenzuela NC 0200-+-+-+-+-+-+-+-")
```
